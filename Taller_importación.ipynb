**Punto 8**

```python
divipola = pd.read_excel(
'/content/drive/MyDrive/GEIH/2025-01-14 DIVIPOLA.xlsx',
sheet_name='Departamentos',
skiprows=9
)
```

```python
divipola.columns
```

```python
divipola = divipola.rename(columns={"Código": "DPTO","Nombre": "DEPARTAMENTO"})
```

```python
divipola = divipola[["DPTO","DEPARTAMENTO"]]
```

```python
df_GEIH_2025['DPTO'] = pd.to_numeric(df_GEIH_2025['DPTO'], errors='coerce')
divipola['DPTO'] = pd.to_numeric(divipola['DPTO'], errors='coerce')
```

```python
df_GEIH_2025 = df_GEIH_2025.merge(
divipola,
on='DPTO',
how='left'
)
```
